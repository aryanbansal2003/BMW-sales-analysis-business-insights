```python
#step 1: import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

#step 2: load data
df=pd.read_csv("BMW.csv")
#step 3: change the indext to start from 1
df.index=df.index+1

#step 4: Understanding The Data
print("A. Understanding The Data")
print("="*88)
print("A.1 Total Number of Records in Dataset:")
print("="*88)
total_records = len(df)
print(total_records)
print("="*88)
print("A.2 Years are covered in the dataset:")
print("="*88)
years = df['Year'].unique()
print(years)
print("="*88)
print("A.3 Unique Models are Present:")
print("="*88)
Unique_Model=df['Model'].unique()
print(Unique_Model)
print("="*88)
print("A.4 Regions where BMW Deal:")
print("="*88)
Region=df['Region'].unique()
print(Region)
print("="*88)
print("A.5 Different Fuel Types:")
print("="*88)
Fuel_Type=df['Fuel_Type'].unique()
print(Fuel_Type)
print("="*88)
print("A.6 Different Transmission Types:")
print("="*88)
Transmission=df['Transmission'].unique()
print(Transmission)
print("="*88)

# setep 5: Sales Performance Analysis
print("="*88)
print("B.Sales Performance Analysis")
print("="*88)
print(df.groupby('Year')['Sales_Volume'].sum())
print("="*88)
print("B.1 Year Which Recorded the Highest Total Sales:")
print("="*88)
print("2022 = 17920946")
print("="*88)
print("B.2 Year Which Recorded The Lowest Sales:")
print("="*88)
print("2023 = 16268654")
print("="*88)
print("B.3 Overall Trend in Sales from 2010 to 2024 (Increasing, Decreasing, Fluctuating):")
print("="*88)
Overall_Trend=df.groupby('Year')['Sales_Volume'].sum()
Overall_Trend.plot(kind='line')
plt.title("Overall Trend in Sales from 2010 to 2024 (Increasing, Decreasing, Fluctuating")
plt.ylabel("Sales_Volume")
plt.xlabel("Year")
plt.show()
print("Interpretation:")
print("The overall sales trend from 2010 to 2024 is fluctuating. While the early years(2010-2016) show relatively stable performance, the later years experience sharp rises and falls in sales volume. The highest sales are observed in 2022, while major declines appear in 2020 and 2023. This suggests that BMW’s sales performance was influenced by changing market conditions, with periods of both strong growth and sudden decline.")
print("="*88)
print("B.4 Model has the Highest Total Sales Volume:")
print("="*88)
Model_Sales = df.groupby('Model')['Sales_Volume'].sum()
top_model = Model_Sales.idxmax()
top_sales = Model_Sales.max()
print("Top Model:", top_model)
print("Total Sales:", top_sales)
print("="*88)
print("B.5 Model has the Lowest Total Sales Volume:")
print("="*88)
Model_Sales = df.groupby('Model')['Sales_Volume'].sum()
top_model = Model_Sales.idxmin()
top_sales = Model_Sales.min()
print("Top Model:", top_model)
print("Total Sales:", top_sales)
print("="*88)

#step 6: Region-wise Insights
print("="*88)
print("C.Region-wise Insights")
print("="*88)
print("C.1 Region that Contributes the Highest and Lowest Sales Volume:")
print("="*88)
region_sales = df.groupby('Region')['Sales_Volume'].sum()
print(region_sales)
print("Interpretation:- The sales distribution across regions shows that Asia contributes the highest sales with 42974277 , making it the most profitable market for BMW. On the other hand, south america  records the lowest sales with  41551818 , indicating weaker demand. Therefore, BMW should prioritize Asia for marketing and inventory, while adopting targeted strategies to improve sales in south America.")
print("="*88)
print("C.2 Region that has the Highest Average Price of Cars?")
print("="*88)
highest_average=df.groupby('Region')['Price_USD'].mean()
print(highest_average)
print("Interpretation- The region with the highest average price is Asia, with an average price of approximately 75,554.93. This indicates that customers in this region have stronger purchasing power and a higher demand for premium vehicles.In contrast, the Middle East has the lowest average price of around 74,726.79, suggesting that customers in this region are relatively more price-sensitive and prefer affordable models.")
print("="*88)
print("C.4 Are high sales classifications concentrated in specific regions")
print("="*88)
Sales_Classification = df.groupby('Region')['Sales_Classification'].value_counts()
print(Sales_Classification)
print("Interpretation- Yes, high sales are concentrated in specific regions like Asia, which shows stronger demand and better market performance.")
print("="*88)


#step 7: Product & Feature Analysis
print("="*88)
print("D.Product & Feature Analysis")
print("="*88)
print("D.1 Which fuel type (Petrol/Diesel/Hybrid) is most common:")
print("="*88)
fuel_type=df['Fuel_Type'].value_counts()
print(fuel_type)
print("Interpretation-The analysis of fuel types shows that Hybrid vehicles are the most common, with a count of 12,716, indicating a strong preference for fuel-efficient and environmentally friendly options. Petrol and Electric vehicles also have similar demand, with counts of 12,550 and 12,471 respectively, showing balanced customer interest in both traditional and modern technologies. On the other hand, Diesel vehicles have the lowest count of 12,263, suggesting relatively lower preference among customers. Overall, the data indicates a shift towards hybrid and cleaner fuel options.")
print("="*88)
print("D.2 transmission type (Manual/Automatic) is more popular:")
print("="*88)
transmission_type=df['Transmission'].value_counts()
print(transmission_type)
print("Interpretation- The analysis of transmission types shows that Manual transmission is slightly more common, with a count of 25,154, compared to Automatic transmission with 24,846. This indicates that customer preference is almost evenly distributed between both types, with a slight inclination towards manual vehicles. The small difference suggests that while many customers still prefer manual transmission, automatic vehicles are also gaining comparable popularity.")
print("="*88)
print("D.3 fuel type contributes to the highest sales volume:")
print("="*88)
fuel_contri=df.groupby('Fuel_Type')['Sales_Volume'].sum()
print(fuel_contri)
print("Interpretation- Hybrid has the highest sales volume at 64532097 units, followed by petrol 63324154 and electric 63157665, while Diesel stands at 62361818 units. Overall, sales are quite close across all fuel types, indicating balanced demand rather than dominance of a single fuel category.")
print("="*88)
print("D.4 Transmission Type is Associated with Higher Sales:")
print("="*88)
Tranmission_higher_sales=df.groupby('Transmission')['Sales_Volume'].sum()
print(Tranmission_higher_sales)
print("Interpretation-The analysis shows that Mannual transmission is associated with higher sales volume, contributing 127362077 units, compared to Automatic transmission with 126013657 units. This indicates a strong customer preference for automatic vehicles, likely due to greater convenience and ease of driving. The significant difference in sales volume suggests that automatic transmission dominates the market, and BMW should focus more on promoting and producing automatic variants to maximize sales.")
print("="*88)

#step 8: Pricing Analysis
print("="*88)
print("E.Pricing Analysis")
print("="*88)
print("E.1 Average Price of Cars Overall")
print("="*88)
average_price = df["Price_USD"].mean()
print(average_price)
print("="*88)
print("E.2&3 Model which has the Highest and Lowest Average Price?")
print("="*88)
highest_avg_price=df.groupby('Model')['Price_USD'].mean()
print(highest_avg_price)
print("Interpretation - The analysis of average prices across different BMW models shows that the 7 Series has the highest average price of approximately 75,570.20, making it the most premium model among the given options. It is closely followed by the 3 Series (75,566.23) and i8 (75,366.27), indicating that these models also belong to the higher price segment.On the other hand, models like X6 (74,434.60) and X5 (74,708.12) have relatively lower average prices, suggesting they are comparatively more affordable within the BMW lineup.Overall, the difference in average prices across models is not very large, indicating that BMW maintains a relatively consistent premium pricing strategy across its product range, while slightly positioning certain models like the 7 Series as more luxurious and high-end.")
print("="*88)
print("E.4 Do higher-priced cars always have higher sales volume?")
print("="*88)
print(df.groupby('Model')[['Price_USD','Sales_Volume']].mean())
print("Interpretation- No, higher-priced cars do not always have higher sales volume.From the data, prices of models are quite similar (around $74,000–$75,500), but their sales volumes vary. For example, X1 has one of the higher sales volumes (~5121 units) despite not being the most expensive, while models like 3 Series (~5066 units) have lower sales even with comparable pricing.")
print("="*88)


#step 9: Mileage & Engine Analysis
print("="*88)
print("F.Mileage & Engine Analysis")
print("="*88)
print("F.1 The Average Mileage Across all Cars?")
print("="*88)
print(df.groupby('Model')['Mileage_KM'].mean())
print("="*88)
print("F.2 Do cars with lower mileage tend to have higher prices?")
print("="*88)
print(df.groupby('Model')[['Price_USD','Mileage_KM']].mean())
print("Interpretation - Mileage analysis helps understand efficiency preferences. If lower mileage cars have higher prices, it indicates that high-performance vehicles are priced higher despite lower fuel efficiency.")
print("="*88)
print("F.3 Engine Size Range which is Most Common?")
print("="*88)
print(df['Engine_Size_L'].value_counts())
print("="*88)
print("F.4 Do larger engine sizes correspond to higher prices?")
print("="*88)
print(df.groupby('Model')[['Engine_Size_L','Price_USD']].mean())
print("Interpretation - No, larger engine sizes do not consistently correspond to higher prices.From the data, engine sizes are very close (around 3.23L to 3.26L), but prices vary across models. For example, the 7 Series (3.258L) has a high price (~$75,570), but the i3 (3.262L) has a lower price (~$74,800) despite having a similar or slightly larger engine size.")
print("="*88)

#step 10: Sales Classification
print("="*88)
print("G.Sales Classification")
print("="*88)
print("G.1 How many cars fall under “High” vs “Low” sales classification")
print("="*88)
print(df['Sales_Classification'].value_counts())
print("="*88)
print("G.2 Models which Appear Most Frequently in “High” Sales?")
print("="*88)
print(df[df['Sales_Classification']=='High']['Model'].value_counts())
print("="*88)
print("G.3 Regions which have the Highest Number of “High” Sales?")
print("="*88)
print(df[df['Sales_Classification']=='High']['Region'].value_counts())
print("="*88)

#step 10: Combined Analysis
print("="*88)
print("H.Combined Analysis")
print("="*88)
print("H.1 Combination of Model + Region gives the highest sales?")
print("="*88)
print(df.groupby(['Model','Region'])['Sales_Volume'].sum().sort_values(ascending=False).head())
print("Interpretation - This indicates that the BMW i8 performs exceptionally well in the European market, suggesting strong customer demand for this model in that region. BMW should focus on strengthening its presence and marketing strategies for the i8 in Europe to further maximize sales.")
print("="*88)
print("H.2 Combination of Fuel Type + Transmission performs best?")
print("="*88)
result = df.groupby(['Fuel_Type', 'Transmission'])['Sales_Volume'].sum()
print(result)
print("Interpretation - This indicates that customers show the strongest preference for hybrid vehicles with manual transmission. It suggests a growing demand for fuel-efficient cars combined with greater driving control, making this combination the top-performing segment for BMW.")
print("="*88)
print("H.3 segment where sales are low but potential seems high (based on price/features).")
print("="*88)
segment = df.groupby(['Fuel_Type', 'Transmission'])[['Sales_Volume','Mileage_KM']].mean()
print(segment)
print("Interpretation - Petrol is only segment which gives low sales but has high potential (Automatic  5018.587979  100362.292459)")
print("="*88)


#step 11: Visualization-Based Questions 
print("="*88)
print("I.Visualization-Based Questions")
print("="*88)
print("I.1 year-wise sales trends and explain the pattern:")
print("="*88)
Sales_Trend=df.groupby('Year')['Sales_Volume'].sum()
Sales_Trend.plot(kind='line')
plt.title("Year wise Sales Trend")
plt.ylabel("Sales_Volume")
plt.xlabel("Year")
plt.show()
print("Interpretation - The overall sales trend from 2010 to 2024 is fluctuating. While the early years(2010-2016) show relatively stable performance, the later years experience sharp rises and falls in sales volume. The highest sales are observed in 2022, while major declines appear in 2020 and 2023.This suggests that BMW’s sales performance was influenced by changing market conditions, with periods of both strong growth and sudden decline.")
print("="*88)
print("I.2 bar chart of sales by region")
print("="*88)
Sales_by_region=df.groupby('Region')['Sales_Volume'].sum()
Sales_by_region.plot(kind='bar',color=['green','blue','yellow','purple','pink','skyblue','red','grey','orange','violet','brown'])
plt.title("Sales by region")
plt.ylabel("Sales_Volume")
plt.xlabel("Region")
plt.show()
print("Interpretation - The bar chart shows a clear variation in sales across regions. One or two regions contribute significantly higher sales compared to others, indicating strong market demand and better performance in those areas. In contrast, some regions have relatively lower sales, suggesting untapped potential or weaker market penetration.")
print("="*88)
print("I.3 price vs mileage:")
print("="*88)
plt.scatter(df['Mileage_KM'], df['Price_USD'])
plt.xlabel("Mileage")
plt.ylabel("Price (USD)")
plt.title("Price vs Mileage")
plt.grid()
plt.show()
print("="*88)
print("I.4 Visualize sales by model and identify top performers")
print("="*88)
Sales_Model=df.groupby('Model')['Sales_Volume'].sum()
Sales_Model.plot(kind='bar',color=['green','blue','yellow','purple','pink','skyblue'])
plt.title("Sales by Model")
plt.ylabel("Sales_Volume")
plt.xlabel("Model")
plt.show()
print("Interpretation - The visualization of sales by model shows that a few BMW models contribute significantly higher sales compared to others. These top-performing models indicate strong customer preference and market demand, while some models have relatively lower sales, suggesting limited popularity or niche demand.This uneven distribution highlights that BMW’s revenue is largely driven by a few key models. The company should focus on promoting high-performing models while also improving the positioning and features of low-performing models to enhance their sales.")
print("="*88)

#step 12: Final Interpretation
print("="*88)
print("J.Final Interpretation")
print("="*88)
print("J.1 If you were a BMW manager, which model would you promote in Asia and why? ")
print("="*88)
print("Answer:- As a BMW manager, I would promote the X1 model in the Asia region, as it has the highest sales volume of 4,192,289 units in this market.This indicates strong customer demand and acceptance of the X1 in Asia. The model likely offers a good balance of price, features, and practicality, making it suitable for a wide range of customers.Increasing marketing efforts and availability of this model can further boost BMW’s sales in the Asian market.")
print("="*88)
print("J.2 What are the top 3 insights we discovered from the dataset?")
print("="*88)
print("Answer:-  1) Sales Trend Insight:The highest sales were recorded in 2022, with 17,920,946 units, showing a peak in demand during this period. 2) Top Model Insight:The 7 Series is the highest-selling model overall, with total sales of 23,786,466 units, indicating strong customer preference for this model. 3) Regional Insight:The Asia region contributes the highest sales volume, with 42,974,277 units, making it BMW’s most important market.")
print("="*88)
print("J.3 3 Business Recommendations based on our Analysis:")
print("="*88)
print("Answer:- 1) Focus on High-Performing Models:BMW should prioritize models like the 7 Series and X1, as they generate the highest sales and drive overall revenue. 2) Strengthen Presence in Asia:Since Asia has the highest sales (42.97 million units), BMW should expand dealerships, improve supply chains, and launch more region-specific models. 3) Develop High-Potential Segments:The Petrol + Automatic segment has relatively lower sales (~5018 units average) but high price (~$75,237), indicating strong potential. BMW should promote this segment through better marketing and positioning.")
print("="*88)
```
